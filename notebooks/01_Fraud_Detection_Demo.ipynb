## 2. Fraud Detection Scoring


```python
import pandas as pd

# Sample transaction features
transaction = pd.DataFrame([{
    'txn_count_1h': 8, 'txn_count_24h': 23, 'txn_count_7d': 89,
    'txn_amount_sum_1h': 4200.0, 'txn_amount_sum_24h': 8900.0,
    'txn_amount_avg_7d': 320.0, 'txn_amount_stddev_7d': 450.0,
    'amount_zscore': 3.8, 'log_amount': 8.2,
    'is_first_merchant_visit': 1, 'is_cross_border': 1,
    'amount_vs_merchant_avg': 4.2, 'hour_of_day': 2,
    'day_of_week': 7, 'is_weekend': 1, 'is_night_txn': 1,
    'seconds_since_last_txn': 45, 'is_round_amount': 1,
}])

model = FraudDetectionModel.load('models/fraud_detector')
fraud_prob = model.predict_proba(transaction)[0]
print(f'Fraud Probability: {fraud_prob:.1%}')
print(f'Is Fraud: {fraud_prob >= model.threshold}')
```

## 3. SHAP Explainability


```python
shap_values = model.explain(transaction, top_n=5)
print('Top 5 Most Important Features:')
print(shap_values)
```
